# A. Introduction
Transit-Oriented Development (TOD) is a city planning approach that focuses on creating mixed-use communities centered around public transportation hubs. Picture neighborhoods where homes, restaurants, shops, and parks are just a short, easy walk—usually 5 to 10 minutes, or around 400 to 800 meters—from a transit station. These areas are thoughtfully designed to reduce the need for cars, making neighborhoods safer and more accessible for people who prefer walking, cycling, or using public transit.

Plus, the Multi-Family Zoning Requirement for MBTA Communities was introduced in January 2021 as a part of Massachusetts' economic development bill. Its goal is to help solve the state's housing shortage. It aims to encourage more multi-family housing near public transit. This rule requires MBTA-served communities to create zoning districts that permit multi-family housing by right. These districts must be close to transit stations.


The TOD approach aligns closely with the goals of the MBTA Communities Zoning Requirement. By emphasizing high-density, multi-family housing in proximity to transit stations, TOD projects directly contribute to alleviating the housing shortage while enhancing accessibility to public transportation. This synergy not only supports the policy's intent to expand housing opportunities but also reinforces sustainability objectives by encouraging public transit use over car dependency, reducing environmental impact, and creating more walkable, connected communities.

For this section, we focus on identifying the best locations for TOD in the great Boston area. By mainly analyzing ridership, convenience and cost-efficiency factors, we aim to support the goals of the MBTA zoning requirement and promote sustainable urban growth.

The site should also support high-density development, including residential, commercial, or mixed-use spaces, to make the best use of the land.

Now we can use the geopy library to find ZIP codes based on latitude and longitude, and append the results as a new column zip_code in the station_pos DataFrame. It may takes for a while.


```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

# Initialize the geolocator
geolocator = Nominatim(user_agent="station_zipcode_locator")

# Add rate limiter to avoid hitting API limits
geocode = RateLimiter(geolocator.reverse, min_delay_seconds=1)

# Function to retrieve ZIP code
def get_zip_code(lat, lon):
    try:
        location = geocode((lat, lon), addressdetails=True)
        if location and 'postcode' in location.raw['address']:
            return location.raw['address']['postcode']
        return None
    except Exception as e:
        print(f"Error retrieving ZIP code for coordinates ({lat}, {lon}): {e}")
        return None
```

```python
# Add a new column for ZIP codes to the station_pos DataFrame
station_pos['zip_code'] = station_pos.apply(
    lambda row: get_zip_code(row['position_y'], row['position_x']), axis=1
)

station_pos.loc[station_pos['stop_name'] == 'Harvard', 'zip_code'] = '02138'

from IPython.display import display
print("Station Data with ZIP Codes") # Print the name
display(station_pos) # Display the dataframe
```

The entry and exit data of 45 stations are summed up according to the weekdays, Saturdays and Sundays.

```python
# Group by 'station_name' and 'day_type' (weekday, saturday, sunday), then sum 'total_ons' and 'total_offs'
aggregated_rider_data = (
    candidate_rider.groupby(['stop_name', 'day_type_name'])[['total_ons', 'total_offs']]
    .sum()
    .reset_index()
)

aggregated_rider_data
```

Calculate the average passenger flow in three cases and separate them into three dataframes.

```python
# Define the number of days for each day type
days_count = {'Weekday': 77, 'Saturday': 16, 'Sunday': 16}


# Add average columns for each day type
aggregated_rider_data['avg_ons'] = aggregated_rider_data.apply(
    lambda row: row['total_ons'] / days_count[row['day_type_name']], axis=1
)
aggregated_rider_data['avg_offs'] = aggregated_rider_data.apply(
    lambda row: row['total_offs'] / days_count[row['day_type_name']], axis=1
)

# Format avg_ons and avg_offs to three decimal places
aggregated_rider_data['avg_ons'] = aggregated_rider_data['avg_ons'].round(3)
aggregated_rider_data['avg_offs'] = aggregated_rider_data['avg_offs'].round(3)

# Split the DataFrame based on 'day_type_name'
wkd_rider = aggregated_rider_data[aggregated_rider_data['day_type_name'] == 'Weekday']
sat_rider = aggregated_rider_data[aggregated_rider_data['day_type_name'] == 'Saturday']
sun_rider = aggregated_rider_data[aggregated_rider_data['day_type_name'] == 'Sunday']

# Add a column for total days corresponding to each day type
wkd_rider['total_days'] = days_count['Weekday']
sat_rider['total_days'] = days_count['Saturday']
sun_rider['total_days'] = days_count['Sunday']

# Concatenate all riders into one DataFrame
all_rider = pd.concat([wkd_rider, sat_rider, sun_rider])

# Group by station and calculate the totals and averages
tot_rider = all_rider.groupby('stop_name').apply(
    lambda group: pd.Series({
        'total_ons': group['total_ons'].sum(),
        'total_offs': group['total_offs'].sum(),
        'avg_ons': group['total_ons'].sum() / group['total_days'].sum(),
        'avg_offs': group['total_offs'].sum() / group['total_days'].sum(),
    })
).reset_index()

# Format avg_ons and avg_offs to three decimal places
tot_rider['avg_ons'] = tot_rider['avg_ons'].round(3)
tot_rider['avg_offs'] = tot_rider['avg_offs'].round(3)
```

# J. Modeling

To determine the most suitable locations for Transit-Oriented Development (TOD) projects, I will use a two-step modeling framework to identify the best locations for Transit-Oriented Development (TOD) projects. The first step will evaluate and rank candidate communities using key indicators. These include **inbound rate, transit coverage, developable station area, community commercial land use, and land price**. This method ensures that the chosen community balances economic feasibility, development potential, and transit accessibility.

In the second step, I will narrow the analysis to the station level within the chosen community. Specifically, I will use **ridership** data to make sure the model will identify the station with the highest potential for TOD development, considering factors such as daily passenger flow. This phased approach provides a data-driven foundation for selecting TOD locations that maximize impact and efficiency.

All the dataframes and indicators are ready, now we are starting to create the assess_rtc summary dataframe.

```python
# Merge the dataframes to create access_rtc based on the updated information
assess_rtc = final_rtc[['community']].merge(
    city_avg_ons[['city', 'avg_ons_percentage']].rename(columns={'city': 'community'}),
    on='community', how='left'
).merge(
    final_rtc[['community', 'percent_of_district_to_be_located_in_station_area']],
    on='community', how='left'
).merge(
    final_rtc[['community', 'developable_station_area']],
    on='community', how='left'
).merge(
    community_land_use_count[['muni', 'count']].rename(columns={'muni': 'community', 'count': 'commercial_land_use'}),
    on='community', how='left'
).merge(
    rtc_avg_price[['municipal', 'avg_price']].rename(columns={'municipal': 'community'}),
    on='community', how='left'
)

# Rename the columns for clarity
assess_rtc.columns = [
    'community', 'inbound_rate', 'coverage', 'developable_station_area',
    'commercial_land_use', 'land_price'
]

# Display the resulting DataFrame
assess_rtc
```

Now that assess_rtc is ready, let's start building the community evaluation model.



---

**STEP 1-1 Preprocess the Data**

1-1-1 Convert percentages in inbound_rate to numeric values.


```python
# Convert percentages in the 'inbound_rate' column to numeric values
assess_rtc['inbound_rate'] = assess_rtc['inbound_rate'].str.rstrip('%').astype(float) / 100

# Display the updated DataFrame
assess_rtc
```

1-1-2 Normalize all indicators to a [0, 1] scale using **Min-Max normalization**.

For community evaluation, we want to combine indicators with different units and ranges into a single score. Min-Max normalization works well in such cases by avoiding dominance by indicators with larger magnitudes, and it is easy to interpret each indicator's contribution relative to others.

```python
# Define the columns to normalize
columns_to_normalize = [
    'inbound_rate', 'coverage', 'developable_station_area',
    'commercial_land_use', 'land_price'
]

# Create a copy of the DataFrame to apply normalization
normalized_rtc = assess_rtc.copy()

# Perform Min-Max normalization
for column in columns_to_normalize:
  normalized_rtc[column] = (normalized_rtc[column] - normalized_rtc[column].min()) / (normalized_rtc[column].max() - normalized_rtc[column].min())

# Display the normalized DataFrame
normalized_rtc
```

In order to ensure that the difference is not exaggerated, a small adjustment is introduced here to smooth the extreme values.

![Min-max normalisation with a smoothing term](images/minmax-smoothing.jpg)

Here, I will use **epsilon = 1e-5** to prevent extremes from being exactly 0 or 1.


```python
# Apply Min-Max normalization with smoothing for the 'coverage' column and others
epsilon = 1e-5  # Smoothing factor to avoid extreme values of 0 and 1

# Copy the assess_rtc DataFrame for normalized values
smoothed_normalized_rtc = assess_rtc.copy()

# Perform Min-Max normalization with smoothing for all indicators
for column in ['inbound_rate', 'coverage', 'developable_station_area', 'commercial_land_use', 'land_price']:
  # epsilon on both sides of the fraction. With it in the denominator alone the
  # maximum lands just under 1 but the minimum is still exactly 0, which is
  # half of what the smoothing is for.
  lo = smoothed_normalized_rtc[column].min()
  hi = smoothed_normalized_rtc[column].max()
  smoothed_normalized_rtc[column] = (
      smoothed_normalized_rtc[column] - lo + epsilon) / (hi - lo + 2 * epsilon)

# Display the resulting DataFrame
smoothed_normalized_rtc
```

1-1-3 Adjustments for "Lower is Better" Indicators

* **commercial_land_use**: Less existing commercial infrastructure means increaseing potential for impactful development.

* **land_price**: Lower land prices reduce development costs, making TOD more financially feasible.

```python
# Perform normalization with inversion for 'commercial_land_use' and 'land_price'
inverse_columns = ['commercial_land_use', 'land_price']

# Copy the normalized data to apply inversion
inverted_rtc = smoothed_normalized_rtc.copy()

# Apply inverse normalization to the specified columns
for column in inverse_columns:
    inverted_rtc[column] = 1 - smoothed_normalized_rtc[column]

# Display the updated DataFrame
inverted_rtc
```

**STEP 1-2 Assign Weights**

The weights are distributed as follows:

*  **inbound_rate: 20%**

  * I believe understanding how reliant a community already is on public transit is essential. The inbound rate reflects the percentage of residents using transit, indicating the community's current alignment with TOD principles. While it's important, I see it as a supporting factor rather than the main driver for TOD development, which is why I assigned it a weight of 20%.
*  **coverage: 25%**

  * Coverage is one of the most critical factors. It measures how much of the community is within reach of a station area, directly impacting accessibility and TOD effectiveness. Communities with high coverage offer greater opportunities for integrated development. This is why I gave it a significant weight of 25%, emphasizing its importance in the TOD projects.

*  **developable_station_area: 15%**

  * Although important, I view the developable station area—the land available for development near stations—as somewhat secondary. Even communities with smaller station areas can succeed with TOD if they have strong connectivity and other supportive factors. For this reason, I assigned it a moderate weight of 15%.

*  **commercial_land_use: 10%**

  * I consider commercial land use a useful but less critical indicator. A lower level of existing commercial land use suggests more untapped potential for TOD development, but it's not a deciding factor. TOD can adapt to varying levels of commercial infrastructure, so I gave this indicator a lower weight of 10%.

*  **land_price: 30%**

  * In my opinion, land price is one of the most decisive factors for TOD feasibility. Affordable land significantly reduces development costs, making projects more viable and attractive. This is why I assigned it a weight of 30%, highlighting the need to focus on financial feasibility when selecting communities for TOD.

```python
# Define the weights as a dictionary
weights_data = {
    'Indicator': [
        'inbound_rate',
        'coverage',
        'developable_station_area',
        'commercial_land_use',
        'land_price'
    ],
    'Weight': [
        0.20,  # Inbound Rate
        0.25,  # Coverage
        0.15,  # Developable Station Area
        0.10,  # Commercial Land Use
        0.30   # Land Price
    ]
}

# Create the weights DataFrame
weights = pd.DataFrame(weights_data)

# Display the DataFrame
print(weights)
```

**STEP 1-3 Calculate Community Scores**

```python
# Calculate the weighted score for each community and add it as a new column named 'scores'
inverted_rtc['scores'] = ((
    inverted_rtc['inbound_rate'] * weights.loc[weights['Indicator'] == 'inbound_rate', 'Weight'].iloc[0] +
    inverted_rtc['coverage'] * weights.loc[weights['Indicator'] == 'coverage', 'Weight'].iloc[0] +
    inverted_rtc['developable_station_area'] * weights.loc[weights['Indicator'] == 'developable_station_area', 'Weight'].iloc[0] +
    inverted_rtc['commercial_land_use'] * weights.loc[weights['Indicator'] == 'commercial_land_use', 'Weight'].iloc[0] +
    inverted_rtc['land_price'] * weights.loc[weights['Indicator'] == 'land_price', 'Weight'].iloc[0]
) * 100)

# Display the updated DataFrame with scores included
inverted_rtc
```

**STEP 1-4 Rank Communities**

```python
# Sort the communities by their scores in descending order
ranked_rtc = inverted_rtc.sort_values(by='scores', ascending=False)

# Add a rank column
ranked_rtc['rank'] = ranked_rtc['scores'].rank(ascending=False, method='min').astype(int)

# Display the ranked DataFrame
print(ranked_rtc[['community', 'scores', 'rank']])
```

```python
# Sort the DataFrame by scores in descending order
ranked_rtc = ranked_rtc.sort_values(by='scores', ascending=False)

# Create the bar chart
plt.figure(figsize=(10, 6))
bars = plt.bar(ranked_rtc['community'], ranked_rtc['scores'], color='skyblue')

# Highlight the highest-ranked station with a different color
bars[0].set_color('#FF7F7F')

# Add scores as annotations on top of the bars
for bar, score in zip(bars, ranked_rtc['scores']):
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # Center of the bar
        bar.get_height(),                  # Height of the bar
        f"{score:.2f}",                    # Format the score to two decimal places
        ha='center', va='bottom', fontsize=10
    )

# Customize the chart
plt.title('Community Rankings Based on TOD Suitability', fontsize=14)
plt.xlabel('Community', fontsize=12)
plt.ylabel('Score', fontsize=12)
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.tight_layout()

# Show the plot
plt.show()
```

2-1-1 Define the indicators.

In this evaluation session, my idea is to select the average daily passenger flow as the primary indicator, and the secondary indicators are the average daily passenger flow on weekends and two ratios assessing the passenger stability - weekday peak hour flow: off-peak hour flow and weekday average passenger volume: weekend average passenger volume.

**Total Activity**: Captured by daily and weekend ridership.

**Usage Patterns**: Addressed by peak/off-peak and weekday/weekend stability.

**Average Daily Ridership (Primary Metric)**

  * This is the most critical indicator as it reflects the total activity at each station. High daily ridership indicates that the station serves a central role in the transit network, making it a focal point for community interactions.

**Average Weekend Ridership (Secondary Metric)**

  * Weekend ridership highlights a station's relevance beyond commuter traffic, such as leisure or retail activities. Stations with higher weekend usage demonstrate a more balanced and diverse rider base for mixed-use developments.

**Passenger Stability 1 (Peak-to-Off-Peak Ratio)**

  * This ratio compares weekday peak-hour ridership to off-peak ridership. A balanced ratio reflects a station that supports consistent activity throughout the day, making it ideal for TOD.

**Passenger Stability 2 (Weekday-to-Weekend Ratio)**

  * This metric compares average weekday ridership to weekend ridership, providing insights into the station's usage patterns across the week. A low ratio suggests the station is versatile, catering to both commuters and leisure travelers, making it a stronger candidate for TOD.

**Reminder**
This dataset contains detailed ridership data for heavy rail and light rail during the **Fall 2023** rating ***(August 27-December 16, 112 days in total)***.
  * For this period, **non-holiday weekdays** were 80 weekdays - 3 holidays (Labor Day, Columbus Day, Thanksgiving) = **77 days**

  * **Saturdays** were **16 days** in total (Veterans Day falls on a Saturday, so no adjustment).

  * **Sunday** were also **16 days** in total.

```python
# Constants
total_weekday = 77
total_weekend = 32  # 16 Saturdays + 16 Sundays
total_days = total_weekday + total_weekend

# Filter weekday and weekend data
weekday_rider_data = quincy_rider_data[quincy_rider_data['day_type_name'] == 'Weekday']
weekend_rider_data = quincy_rider_data[quincy_rider_data['day_type_name'].str.contains('Saturday|Sunday', case=False)]

# 1. Calculate Average Daily Ridership
daily_avg_ridership = quincy_rider_data.groupby('stop_name')[['total_ons', 'total_offs']].sum()
daily_avg_ridership['avg_daily_ridership'] = (
    daily_avg_ridership['total_ons'] + daily_avg_ridership['total_offs']
) / total_days

# 2. Calculate Average Weekend Ridership
weekend_avg_ridership = weekend_rider_data.groupby('stop_name')[['total_ons', 'total_offs']].sum()
weekend_avg_ridership['avg_weekend_ridership'] = (
    weekend_avg_ridership['total_ons'] + weekend_avg_ridership['total_offs']
) / total_weekend

# 3. Calculate Passenger Stability 1: Weekday Peak Share
# The dataset cuts a weekday into nine time periods, two of which are peaks, so
# peak flow divided by off-peak flow is two buckets against seven. It carries
# no natural reference value, and every station comes out below 1 for a reason
# that has nothing to do with how spread out its day is. The share of the
# weekday that falls inside the peaks is the quantity this indicator is really
# about, it is bounded between 0 and 1, and it is comparable across stations.
# Low is good: a station busy all day can host shops and housing, a station
# busy for two periods cannot.
is_peak = weekday_rider_data['time_period_name'].str.contains(
    'AM_PEAK|PM_PEAK', case=False, na=False)

weekday_flow = weekday_rider_data.groupby('stop_name')[['total_ons', 'total_offs']].sum().sum(axis=1)
peak_flow = weekday_rider_data[is_peak].groupby('stop_name')[['total_ons', 'total_offs']].sum().sum(axis=1)

stability_1 = pd.DataFrame({
    'avg_peak_ridership': peak_flow / total_weekday,
    'peak_share': peak_flow / weekday_flow,
})

# 4. Calculate Passenger Stability 2: Weekday to Weekend Ratio
weekday_totals = weekday_rider_data.groupby('stop_name')[['total_ons', 'total_offs']].sum()
weekday_totals['avg_weekday_ridership'] = (
    weekday_totals['total_ons'] + weekday_totals['total_offs']
) / total_weekday

stability_2 = weekday_totals[['avg_weekday_ridership']].copy()
stability_2['weekday_to_weekend_ratio'] = stability_2['avg_weekday_ridership'] / weekend_avg_ridership['avg_weekend_ridership']

# Merge all metrics into a single DataFrame
metrics_df = daily_avg_ridership[['avg_daily_ridership']].merge(
    weekend_avg_ridership[['avg_weekend_ridership']],
    left_index=True, right_index=True, how='left'
).merge(
    stability_1[['peak_share']],
    left_index=True, right_index=True, how='left'
).merge(
    stability_2[['weekday_to_weekend_ratio']],
    left_index=True, right_index=True, how='left'
)

# Display the combined metrics DataFrame
metrics_df
```

2-1-2 Normalization.

All four indicators are min-max normalised across the four stations, so every column lands on the same 0 to 1 scale and the weights below mean what they say.

Two of them run the other way, and are inverted after scaling. `peak_share` is the fraction of weekday flow that arrives in the two peak periods: the lower it is, the more evenly the station is used across the day. `weekday_to_weekend_ratio` compares an average weekday against an average weekend day, and every station is above 2, so the lower it is the more life the station has at the weekend. For a site meant to carry shops and housing, low is good in both cases.

```python
# Min-max every indicator, then invert the two where low is good. One scale,
# bounded to [0, 1], and no term that can leave it.
normalized_metrics_df = metrics_df.copy()

higher_is_better = ['avg_daily_ridership', 'avg_weekend_ridership']
lower_is_better = ['peak_share', 'weekday_to_weekend_ratio']

for metric in higher_is_better + lower_is_better:
    min_value = metrics_df[metric].min()
    max_value = metrics_df[metric].max()
    scaled = (metrics_df[metric] - min_value) / (max_value - min_value)
    normalized_metrics_df[metric] = 1 - scaled if metric in lower_is_better else scaled

# Display the normalized metrics DataFrame
normalized_metrics_df
```

**STEP 2-2 Assign Weights**

The weights are distributed as follows:

*  **avg_daily_ridership: 40%**

  * I believe that average daily ridership is the most critical factor in ranking Quincy stations, as it directly measures the total activity and the station's importance within the transit network. This is why I have assigned it the highest weight of 40%.

*  **avg_weekend_ridership: 20%**

  * Average weekend ridership, while secondary, is still essential as it reflects the station's relevance beyond weekday commuter traffic. This helps identify stations that cater to a broader range of activities, such as leisure and shopping. I have allocated it a weight of 20%, acknowledging its role in assessing the station's versatility.

*  **peak_share: 20%**

  * Passenger stability 1 (peak-to-off-peak ratio) is another important indicator, receiving a weight of 20%. A balanced ratio reflects consistent activity throughout the day, indicating a station's potential to support TOD.

*  **weekday_to_weekend_ratio: 20%**

  * Passenger stability 2 (weekday-to-weekend ratio) is equally weighted at 20%. This metric ensures that stations with diverse rider profiles, serving both commuters and non-commuters, are given due consideration. A balanced weekday-to-weekend ratio aligns well with the TOD goal of creating vibrant, multi-purpose areas.

```python
# Define weights for each indicator
weights_stations = {
    'avg_daily_ridership': 0.40,
    'avg_weekend_ridership': 0.20,
    'peak_share': 0.20,
    'weekday_to_weekend_ratio': 0.20
}

# Create the weights DataFrame
weights_station = pd.DataFrame([weights_stations]).T
# Set the column name
weights_station.columns = ['Weight']

# Display the DataFrame
weights_station
```

**STEP 2-3 Calculate Station Scores**

```python
# Calculate the weighted score for each station
# No offset. Every normalised column is now inside [0, 1], so the weighted sum
# is already on a 0 to 100 scale and the flat +20 that used to lift the scores
# off the floor has nothing left to correct.
normalized_metrics_df['final_score'] = ((
    normalized_metrics_df['avg_daily_ridership'] * weights_stations['avg_daily_ridership'] +
    normalized_metrics_df['avg_weekend_ridership'] * weights_stations['avg_weekend_ridership'] +
    normalized_metrics_df['peak_share'] * weights_stations['peak_share'] +
    normalized_metrics_df['weekday_to_weekend_ratio'] * weights_stations['weekday_to_weekend_ratio']
) * 100)

# Sort the stations by their final score in descending order
ranked_stations = normalized_metrics_df.sort_values(by='final_score', ascending=False)

ranked_stations
```

```python
# Sort the stations by their final score in descending order
ranked_stations = normalized_metrics_df.sort_values(by='final_score', ascending=False)

# Display the ranked stations DataFrame
ranked_stations
```
